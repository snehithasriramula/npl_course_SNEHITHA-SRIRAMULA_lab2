```python
import os, re, urllib.request
import nltk
import pandas as pd
import numpy as np
for resource in ["punkt", "punkt_tab", "stopwords", "wordnet"]:
    nltk.download(resource, quiet=True)
from nltk.tokenize import word_tokenize, sent_tokenize
from nltk.stem import PorterStemmer, WordNetLemmatizer
from nltk.corpus import stopwords

def local_file(name):
    path = os.path.join("data", name)
    if not os.path.exists(path):
        os.makedirs("data", exist_ok=True)
        urllib.request.urlretrieve(f"https://raw.githubusercontent.com/ababacaryoro/nlp-course/main/assignments/1/data/{name}", path)
    return path

with open(local_file("moby.txt"), encoding="utf-8") as f:
    moby_raw = f.read()
moby_tokens = word_tokenize(moby_raw)
text1 = nltk.Text(moby_tokens)
print(f"{len(moby_raw)} characters, {len(moby_tokens)} tokens")
print(moby_raw[:200])
```

## Examples

```python
def example_one():
    """Number of tokens."""
    return len(moby_tokens)

def example_two():
    """Number of distinct tokens."""
    return len(set(moby_tokens))

def example_three():
    """Number of distinct tokens after lemmatizing verbs."""
    lemmatizer = WordNetLemmatizer()
    return len(set(lemmatizer.lemmatize(w, "v") for w in text1))

print(example_one(), example_two(), example_three())
```

## Question 1 (1 point)

Lexical diversity: number of distinct tokens divided by the number of tokens. *Returns a float.*

```python
def question_one():
    """Lexical diversity: distinct tokens divided by total tokens."""
    if not moby_tokens:
        return 0.0
    return len(set(moby_tokens)) / len(moby_tokens)

print(question_one())   # about 0.08
```

## Question 2 (1 point)

Percentage of tokens equal to `whale` or `Whale`. *Returns a float (e.g. 0.5 for 0.5 %).*

```python
def question_two():
    """Percentage of tokens equal to 'whale' or 'Whale'. E.g. 0.5 means 0.5 %."""
    if not moby_tokens:
        return 0.0
    count = sum(1 for t in moby_tokens if t in ("whale", "Whale"))
    return 100 * count / len(moby_tokens)

print(question_two())
```

## Question 3 (1 point)

The 20 most frequent tokens and their frequency. *Returns a list of 20 `(token, frequency)` tuples sorted by decreasing frequency.*

```python
def question_three():
    """The 20 most frequent tokens and their frequency."""
    fdist = nltk.FreqDist(moby_tokens)
    return fdist.most_common(20)

print(question_three())
```

## Question 4 (1 point)

Tokens longer than 5 characters with a frequency above 150. *Returns an alphabetically sorted list of tokens.*

```python
def question_four():
    """Tokens longer than 5 characters with a frequency above 150."""
    fdist = nltk.FreqDist(moby_tokens)
    filtered_tokens = [token for token, freq in fdist.items() if len(token) > 5 and freq > 150]
    return sorted(filtered_tokens)

print(question_four())
```

## Question 5 (1 point)

The longest token and its length. *Returns a `(token, length)` tuple.*

```python
def question_five():
    """The longest token and its length."""
    longest = max(moby_tokens, key=len)
    return (longest, len(longest))

print(question_five())
```

## Question 6 (1 point)

Alphabetic tokens (`isalpha()`) with a frequency above 2000. *Returns a list of `(frequency, word)` tuples sorted by decreasing frequency.*

```python
def question_six():
    """Alphabetic tokens (isalpha()) with a frequency above 2000."""
    fdist = nltk.FreqDist(moby_tokens)
    # Filter and format as (frequency, word) for alphabetic tokens with frequency > 2000
    filtered = [
        (freq, token)
        for token, freq in fdist.items()
        if token.isalpha() and freq > 2000
    ]
    # Sort in descending order of frequency
    return sorted(filtered, key=lambda x: x[0], reverse=True)

print(question_six())
```

## Question 7 (1 point)

Average number of tokens per sentence (`sent_tokenize` for the sentences). *Returns a float.*

```python
def question_seven():
    """Average number of tokens per sentence."""
    sentences = sent_tokenize(moby_raw)
    if not sentences:
        return 0.0
    return len(moby_tokens) / len(sentences)

print(question_seven())
```

## Question 8 (1 point)

The 10 most frequent lowercase alphabetic tokens after removing the English stop words. *Returns a list of 10 `(word, frequency)` tuples sorted by decreasing frequency.*

```python
def question_eight():
    """The 10 most frequent lowercase alphabetic tokens after removing the English stop words."""
    stop_words = set(stopwords.words("english"))
    # Lowercase, keep alphabetic, and filter out stop words
    filtered_tokens = [
        token.lower()
        for token in moby_tokens
        if token.isalpha() and token.lower() not in stop_words
    ]
    fdist = nltk.FreqDist(filtered_tokens)
    return fdist.most_common(10)

print(question_eight())
```

## Question 9 (1 point)

The 10 most frequent Porter stems of the alphabetic tokens. *Returns a list of 10 `(stem, frequency)` tuples sorted by decreasing frequency.*

```python
def question_nine():
    """The 10 most frequent Porter stems of the alphabetic tokens."""
    stemmer = PorterStemmer()
    # Filter to keep only alphabetic tokens and apply Porter stemmer
    stems = [stemmer.stem(token.lower()) for token in moby_tokens if token.isalpha()]
    fdist = nltk.FreqDist(stems)
    return fdist.most_common(10)

print(question_nine())
```

## Question 10 (1 point)

Preprocess the first 1 000 characters of the novel: tokenize, lowercase, keep alphabetic tokens, remove stop words, lemmatize (default noun POS). *Returns the list of tokens.*

```python
def question_ten():
    text = moby_raw[:1000]
    stop_words = set(stopwords.words("english"))
    lemmatizer = WordNetLemmatizer()

    # Tokenize the 1,000 characters
    tokens = word_tokenize(text)

    # Lowercase, keep alphabetic, remove stop words, and lemmatize
    preprocessed_tokens = [
        lemmatizer.lemmatize(token.lower())
        for token in tokens
        if token.isalpha() and token.lower() not in stop_words
    ]
    return preprocessed_tokens

r = question_ten()
print(len(r), r[:20])
```
